# Layer Normalization Interview Questions — Answers

---

# 1. What problem does Layer Normalization solve in deep learning?

LayerNorm stabilizes training by normalizing activations so they have:
- approximately zero mean
- approximately unit variance

This reduces internal covariate shift and improves gradient flow.

It helps:
- faster convergence
- more stable optimization
- deeper network training

---

# 2. Difference between BatchNorm, LayerNorm, InstanceNorm, GroupNorm

| Method | Normalizes Across | Depends on Batch? | Common Usage |
|---|---|---|---|
| BatchNorm | Batch dimension | Yes | CNNs |
| LayerNorm | Feature dimension | No | Transformers, RNNs |
| InstanceNorm | Spatial dimensions per channel | No | Style transfer |
| GroupNorm | Groups of channels | No | Small-batch CNNs |

LayerNorm is preferred over BatchNorm when:
- batch sizes are small
- sequence lengths vary
- autoregressive models are used
- transformers are trained

---

# 3. Why is LayerNorm used in Transformers instead of BatchNorm?

Because:
- Transformer tokens are processed independently
- Batch statistics are unstable for variable sequence lengths
- Autoregressive inference often uses batch size = 1

LayerNorm does not depend on batch statistics.

---

# 4. Along which dimensions is normalization performed?

For:

```python
x.shape = (batch_size, seq_len, hidden_dim)
```

LayerNorm usually normalizes over:

```python
hidden_dim
```

meaning:

```python
dim = -1
```

Each token is normalized independently.

- normalization may remove useful scale information

# 11. Implement LayerNorm from scratch

```python
def my_layer_norm(x, gamma, beta, eps=1e-5):
    mean = x.mean(dim=-1, keepdim=True)
    var = x.var(dim=-1, keepdim=True, unbiased=False)

    x_hat = (x - mean) / torch.sqrt(var + eps)

    return gamma * x_hat + beta
```

---

# 12. What is wrong here?

```python
var = x.var(dim=-1, keepdim=True)
```

Problem:
- default uses `unbiased=True`
- LayerNorm requires population variance

Fix:

```python
unbiased=False
```

---

# 13. Why use keepdim=True?

Without it:

```python
mean.shape = (batch, seq)
```

With it:

```python
mean.shape = (batch, seq, 1)
```

This allows broadcasting during:

```python
x - mean
```

# 16. Why better for small batch sizes?

BatchNorm relies on stable batch statistics.

Small batches produce noisy estimates.

LayerNorm works independently per sample.

---

# 17. Why independent of batch statistics?

Normalization is performed within each sample.

Benefits:
- stable inference
- works with batch size 1
- works in autoregressive decoding

---

# 18. Pre-LN vs Post-LN

Pre-LN:

```python
LN -> Attention -> Residual
```

Post-LN:

```python
Attention -> Residual -> LN
```

Pre-LN:
- better gradient flow
- more stable deep transformers

Modern LLMs mostly use Pre-LN.

---

# 19. How does LayerNorm affect gradient flow?

It:
- stabilizes activation magnitudes
- prevents exploding/vanishing gradients
- smooths optimization landscape

---

# 20. Can LayerNorm replace BatchNorm in CNNs?

Sometimes, but usually BatchNorm performs better in CNNs because:
- convolutional features benefit from batch statistics
- spatial correlations are important

LayerNorm is more common in:
- transformers
- NLP
- sequence models
