Implement the **Adam** optimizer from scratch.

### Signature
```python
class MyAdam:
    def __init__(self, params, lr=1e-3, betas=(0.9, 0.999), eps=1e-8): ...
    def step(self): ...
    def zero_grad(self): ...
```

### Algorithm (per parameter)
```
m = β1 * m + (1-β1) * grad
v = β2 * v + (1-β2) * grad²
m̂ = m / (1 - β1ᵗ)    # bias correction
v̂ = v / (1 - β2ᵗ)
p -= lr * m̂ / (√v̂ + ε)
```

````python
# ✏️ YOUR IMPLEMENTATION HERE

class MyAdam:
    def __init__(self, params, lr=1e-3, betas=(0.9, 0.999), eps=1e-8):
        self.params=params
        self.m=[torch.zeros_like(p) for p in params]
        self.v=[torch.zeros_like(p) for p in params]
        self.lr=lr
        self.eps=eps
        self.betas=betas
        self.t=0
    def step(self):
        self.t+=1
        with torch.no_grad():

          for  i,p in enumerate(self.params):
            if p.grad is None:
              continue



            self.m[i]=self.betas[0]*self.m[i]+(1-self.betas[0])*p.grad
            self.v[i]= self.betas[1]*self.v[i]+(1-self.betas[1])*p.grad**2
            m_prime=self.m[i]/(1-self.betas[0]**self.t)
            v_prime=self.v[i]/(1-self.betas[1]**self.t)
            self.params[i]= self.params[i]- self.lr*m_prime/(torch.sqrt(v_prime)+self.eps)
            #print(self.lr*m_prime/(torch.sqrt(v_prime)+self.eps))
            print(self.params[i])
            #print(self.params[i].grad)

        pass  # update params using Adam rule

    def zero_grad(self):
        for i,p in  enumerate (self.params) :
          self.params[i].grad=torch.zeros_like(p.grad)
          #p.grad=torch.zeros_like(p.grad)
        pass  # zero all gradients
````

````python
# 🧪 Debug
torch.manual_seed(0)
w = torch.randn(4, 3, requires_grad=True)
opt = MyAdam([w], lr=0.01)
for i in range(5):
    loss = (w ** 2).sum()
    loss.backward()
    opt.step()
    opt.zero_grad()
    print(f'Step {i}: loss={loss.item():.4f}')
````

## Answers

### Warm-up

**1. What problem does Adam solve compared to vanilla SGD?**  
Adam adapts the learning rate for each parameter individually using gradient statistics, leading to faster and more stable convergence.

**2. Why two moving averages (m and v)?**  
m tracks the mean of gradients (momentum), while v tracks the variance (scale). Together they give direction and adaptive step size.

**3. What do beta1 and beta2 control?**  
beta1 controls momentum smoothing. beta2 controls variance smoothing.

---

### Core Mechanics

**4. Full Adam update equations:**

```
m_t = beta1 * m_{t-1} + (1 - beta1) * g_t
v_t = beta2 * v_{t-1} + (1 - beta2) * g_t^2

m_hat = m_t / (1 - beta1^t)
v_hat = v_t / (1 - beta2^t)

theta = theta - lr * m_hat / (sqrt(v_hat) + eps)
```

**5. Why divide by sqrt(v_hat)?**  
Because v_hat estimates squared gradients. Taking square root brings it back to gradient scale.

**6. Effect of eps:**  
Too large makes Adam behave like SGD. Too small risks numerical instability.

**7. Why bias correction?**  
Moving averages start at zero and are biased early. Correction removes this bias.

---

### Implementation-Level

**8. Why torch.no_grad()?**  
To prevent updates from being tracked in the computation graph.

**9. Difference between p = p - update and p -= update?**  
p = p - update creates a new tensor. p -= update modifies the existing tensor in-place.

**10. Why check p.grad is None?**  
Some parameters may not receive gradients. This avoids errors.

---

### Deeper Understanding

**11. Relation to Momentum and RMSProp:**  
Momentum corresponds to m. RMSProp corresponds to v. Adam combines both.

**12. If beta1 = 0 or beta2 = 0:**  
beta1 = 0 removes momentum. beta2 = 0 removes adaptive scaling.

**13. Why worse generalization?**  
Adaptive learning rates can lead to sharper minima compared to SGD.

---

### Practical + Research-Level

**14. AdamW vs Adam:**  
AdamW decouples weight decay from gradient updates.

**15. Why weight decay tricky?**  
In Adam, L2 regularization interacts with adaptive scaling.

**16. When prefer SGD?**  
When generalization is more important, especially in large-scale vision tasks.

---

### Edge Cases / Debugging

**17. Loss not decreasing:**  
Check learning rate, gradients, bias correction, and parameter updates.

**18. Sparse gradients:**  
Adam handles sparse gradients better due to per-parameter scaling.

**19. Why unstable?**  
Large learning rates or poor beta settings can cause instability.

---

### Bonus

**20. Adam as momentum + adaptive LR:**  
Momentum provides direction. Variance scaling adjusts step size.

**21. Diagonal preconditioner:**  
Adam rescales each parameter independently, equivalent to a diagonal matrix scaling gradients.
